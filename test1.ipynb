Find all datasets automatically

```python
repo_path = "/content/ANLOK-WATER-PREDICTION-PROJECT-"

csv_files = glob.glob(
    repo_path + "/**/*.csv",
    recursive=True
)

len(csv_files)
```

Display Datasets

```python
for file in csv_files:
    print(file)
```

Dictionary of all the CSV files

```python
import pandas as pd
import os

datasets = {}

for file in csv_files:
    name = os.path.basename(file).replace(".csv","")
    try:
        datasets[name] = pd.read_csv(file)
    except pd.errors.EmptyDataError:
        print(f"Skipping empty or malformed file: {file}")
        continue # Skip to the next file

print(datasets.keys())
```

inventory

```python
inventory = []

for category, df in datasets.items():

    inventory.append({

            "Category": category,
            "Dataset": category,
            "Rows": df.shape[0],
            "Columns": df.shape[1],
            "Missing Values": df.isnull().sum().sum(),
            "Duplicate Rows": df.duplicated().sum()

        })

inventory_df = pd.DataFrame(inventory)

inventory_df
```

Save inventory

```python
inventory_df.to_csv(
    "ANLOK_dataset_inventory.csv",
    index=False
)
```
